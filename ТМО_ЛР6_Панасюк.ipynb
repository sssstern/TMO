Цель лабораторной работы: 
1. Исследование необходимых и достаточных условий существования
экстремума функции c учетом ограничений (условный экстремум).
2. Вычисление экстремумов функции.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize

data_all = {
    '1.1': [lambda x: (x[0]-2)**2 + (x[1]-3)**2 + x[0]*x[1], 
            [lambda x: x[0]**2 + x[1]**2 - 52, lambda x: -x[0]]],
    
    '1.2': [lambda x: (x[0]**2 - x[1])**2 + x[1]**2, 
            [lambda x: x[0]**2 + x[1]**2 - 1, lambda x: -x[0]]],
    
    '1.3': [lambda x: x[0]**2 + 2*x[0]*x[1] + (x[1]-2)**2, 
            [lambda x: x[0]**2 + x[1]**2 - 1, lambda x: -x[0], lambda x: -x[1]]],
    
    '1.4': [lambda x: 3*x[0]**2 + 4*x[0]*x[1] + 5*x[1]**2, 
            [lambda x: 4 - x[0] - x[1], lambda x: -x[0], lambda x: -x[1]]],
    
    '1.5': [lambda x: 4*x[0]**2 + 4*x[0] + x[1]**2 - 8*x[1] + 5, 
            [lambda x: 2*x[0] + x[1] - 6, lambda x: -x[0], lambda x: -x[1]]],
    
    '1.6': [lambda x: (x[0]-x[1])**2 + (x[1]+2)**2 - 10, 
            [lambda x: x[0]**2 - x[1] - 1, lambda x: -x[0], lambda x: -x[1]]],
    
    '1.7': [lambda x: (x[0])**2 + (x[1]**2 + 2)**2 - 10, 
            [lambda x: x[0]**2 + x[1]**2 - 1, lambda x: -x[0], lambda x: -x[1]]],
    
    '1.8': [lambda x: (x[0]**2 - 5)**2 + (x[1] + 2)**2 - 10, 
            [lambda x: x[0]**2 + 5*x[1]**2 - 1, lambda x: -x[0], lambda x: -x[1]]],
    
    '1.9': [lambda x: (3*np.exp(-x[0]**2) - 2)**2 + (x[1]-4)**2 - 10, 
            [lambda x: 4*x[0]**2 + x[1]**2 - 1, lambda x: -x[0], lambda x: -x[1]]],
    
    '1.10': [lambda x: (10*np.log(max(x[0], 1e-5)) - 2)**2 + 10*(x[1]**2 + 2)**2 - 10, 
             [lambda x: x[0]**2 + x[1]**2 - 2, lambda x: -x[0], lambda x: -x[1]]],
    
    '1.11': [lambda x: 10*(x[0]**2 - x[1])**2 + x[1]**2, 
             [lambda x: x[0]**2 + x[1]**2 - 4, lambda x: -x[0]]],
    
    '1.12': [lambda x: 2*(x[0]**4 - x[1]**2)**2 + 3*x[0]*x[1], 
             [lambda x: x[0]**2 + x[1]**2 - 6, lambda x: -x[0]]],
    
    '1.13': [lambda x: (x[0]**6 - x[1]**2)**2 + 5*x[1]**2 - 10*x[0], 
             [lambda x: x[0]**2 + 4*x[1]**2 - 8, lambda x: -x[0]]],
    
    '1.14': [lambda x: (3*np.log(max(x[0], 1e-5)) - np.exp(x[1]))**2 + (x[1]**2 - 1)**2, 
             [lambda x: x[0]**2 + x[1]**2 - 10, lambda x: -x[0]]],
    
    '1.15': [lambda x: 8*(x[0]**2 - x[1])**2 + (x[1]+2)**2 - 10, 
             [lambda x: x[0]**2 - x[1]**2 - 10, lambda x: -x[0], lambda x: -x[1]]],
    
    '1.16': [lambda x: 4*x[0]**2 + 8*x[0]*x[1] + 2*x[1]**2, 
             [lambda x: x[0] + x[1] - 20, lambda x: -x[0], lambda x: -x[1]]]
}


def run_optimization_lab():
    for task_num, components in data_all.items():
        print(f"\n{'='*50}\n Задание № {task_num} \n{'='*50}")
        
        func = components[0]
        g_constraints = components[1]
        
        # Перевод ограничений в формат SciPy (ineq: c(x) >= 0)
        scipy_cons = []
        for g in g_constraints:
            scipy_cons.append({'type': 'ineq', 'fun': lambda x, g_func=g: -g_func(x)})
            
        # Начальная точка
        x_init = np.array([1.0, 1.0]) if task_num in ['1.10', '1.14'] else np.array([0.5, 0.5])
        
        # Поиск условного экстремума
        result = minimize(func, x_init, constraints=scipy_cons, method='SLSQP')
        
        if result.success:
            x_opt = result.x
            f_opt = result.fun
            print(f"[Успешно] Стационарная точка x^e: [{x_opt[0]:.4f}, {x_opt[1]:.4f}]")
            print(f"Значение функции в точке f(x^e): {f_opt:.4f}")
            
            print("Проверка ограничений g_j(x^e) <= 0:")
            for idx, g in enumerate(g_constraints):
                val = g(x_opt)
                status = "Активно (=0)" if abs(val) < 1e-4 else "Пассивно (<0)"
                print(f"  g_{idx+1}(x^e) = {val:.4f} -> {status}")
                
            plot_contour_and_constraints(task_num, func, g_constraints, x_opt)
        else:
            print(f"[Ошибка] Не удалось найти решение для пункта {task_num}: {result.message}")

def plot_contour_and_constraints(task_num, func, constraints, x_opt):
    x1_min, x1_max = x_opt[0] - 3, x_opt[0] + 3
    x2_min, x2_max = x_opt[1] - 3, x_opt[1] + 3
    
    if task_num in ['1.10', '1.14']:
        x1_min = max(0.01, x1_min)
        
    u = np.linspace(x1_min, x1_max, 200)
    v = np.linspace(x2_min, x2_max, 200)
    X1, X2 = np.meshgrid(u, v)
    
    Z = np.zeros_like(X1)
    for i in range(X1.shape[0]):
        for j in range(X1.shape[1]):
            Z[i, j] = func([X1[i, j], X2[i, j]])
            
    plt.figure(figsize=(9, 7))
    
    contours = plt.contour(X1, X2, Z, levels=40, cmap='viridis', alpha=0.6)
    plt.clabel(contours, inline=True, fontsize=8)
    
    for idx, g in enumerate(constraints):
        Z_g = np.zeros_like(X1)
        for i in range(X1.shape[0]):
            for j in range(X1.shape[1]):
                Z_g[i, j] = g([X1[i, j], X2[i, j]])
        
        plt.contour(X1, X2, Z_g, levels=[0], colors=['red', 'blue', 'green', 'magenta'][idx % 4], 
                    linestyles='--', linewidths=2)
        
    plt.plot(x_opt[0], x_opt[1], 'ro', markersize=10, label='Экстремум x^e')
    
    plt.title(f"Визуализация задачи № {task_num} и области ограничений")
    plt.xlabel("x1")
    plt.ylabel("x2")
    plt.grid(True, linestyle=':', alpha=0.5)
    plt.legend(loc='upper right')
    plt.show()

if __name__ == "__main__":
    run_optimization_lab()
```
