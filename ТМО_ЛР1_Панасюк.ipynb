## 1) Текстовое описание набора данных

В качестве набора данных для анализа был выбран датасет "BankChurners" (Прогнозирование оттока клиентов банка)
Этот датасет содержит информацию о клиентах банка и их кредитных картах, а также помечен, ушел клиент из банка (закрыл карту) или нет.
Датасет состоит из одного файла и содержит следующие колонки:

    - CLIENTNUM: Уникальный идентификатор клиента.
    - Attrition_Flag: Целевой признак. Флаг оттока, показывающий, ушел клиент (Attrited Customer) или все еще является активным (Existing Customer).
    - Customer_Age: Возраст клиента (в годах).
    - Gender: Пол клиента (M - мужчина, F - женщина).
    - Dependent_count: Количество иждивенцев клиента.
    - Education_Level: Уровень образования (например, Graduate, High School, Unknown).
    - Marital_Status: Семейное положение (Married, Single, Unknown).
    - Income_Category: Категория дохода (например, Less than $40K, $60K - $80K).
    - Card_Category: Тип карты (Blue, Silver, Gold, Platinum).
    - Months_on_book: Количество месяцев, в течение которых клиент является клиентом банка.
    - Total_Relationship_Count: Количество продуктов банка, которыми пользуется клиент.
    - Months_Inactive_12_mon: Количество месяцев неактивности за последние 12 месяцев.
    - Contacts_Count_12_mon: Количество обращений в контакт-центр за последние 12 месяцев.
    - Credit_Limit: Кредитный лимит по карте (в долларах).
    - Total_Revolving_Bal: Остаток задолженности на конец расчетного периода.
    - Avg_Open_To_Buy: Средний доступный кредитный лимит (Credit_Limit - Total_Revolving_Bal).
    - Total_Amt_Chng_Q4_Q1: Отношение суммы транзакций за 4-й квартал к сумме за 1-й квартал.
    - Total_Trans_Amt: Общая сумма транзакций за последние 12 месяцев.
    - Total_Trans_Ct: Общее количество транзакций за последние 12 месяцев.
    - Total_Ct_Chng_Q4_Q1: Отношение количества транзакций за 4-й квартал к количеству за 1-й квартал.
    - Avg_Utilization_Ratio: Средний коэффициент использования кредитного лимита.

# 4) Информация о корреляции признаков

```python
# Только числовые признаки
corr = data[num_cols + ["Attrition_Flag"]].copy()
# преобразуем целевой признак в 0/1
corr["Attrition"] = (corr["Attrition_Flag"] == "Attrited Customer").astype(int)
corr = corr.drop(columns="Attrition_Flag")

plt.figure(figsize=(14,11))
sns.heatmap(
    corr.corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.4,
    annot_kws={"size":9}
)
plt.title("Матрица корреляций Пирсона", fontsize=16, pad=15)
plt.show()
```

```python
# Вариант А — явно указываем числовые столбцы
numeric_cols = [
    'Customer_Age', 'Dependent_count', 'Months_on_book',
    'Total_Relationship_Count', 'Months_Inactive_12_mon',
    'Contacts_Count_12_mon', 'Credit_Limit', 'Total_Revolving_Bal',
    'Avg_Open_To_Buy', 'Total_Amt_Chng_Q4_Q1', 'Total_Trans_Amt',
    'Total_Trans_Ct', 'Total_Ct_Chng_Q4_Q1', 'Avg_Utilization_Ratio'
]

fig, ax = plt.subplots(1, 3, figsize=(18, 6))

sns.heatmap(data[numeric_cols].corr(method='pearson'),  ax=ax[0],  cmap='coolwarm')
sns.heatmap(data[numeric_cols].corr(method='kendall'), ax=ax[1],   cmap='coolwarm')
sns.heatmap(data[numeric_cols].corr(method='spearman'), ax=ax[2],   cmap='coolwarm')

ax[0].set_title('Pearson')
ax[1].set_title('Kendall')
ax[2].set_title('Spearman')

fig.suptitle('Корреляционные матрицы (только числовые признаки)', y=1.02, fontsize=14)
plt.tight_layout()
plt.show()
```
